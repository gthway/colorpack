```python
from colorpack import ColorSet, shades, gradient, diverging, COLORS, PALETTES, PALETTE_NAMES
import numpy as np
import matplotlib.pyplot as plt
```

```python
print(list(COLORS.keys()))
print(PALETTE_NAMES)
print(PALETTES)
```

```python
shades_of_red = shades('red',5)
redblue_grad = gradient('red','blue',6)
redblue_diverging_grad = diverging('red','blue',6)
```

```python
x = np.arange(0,1,0.1)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = 2*np.sin(x)
y4 = 2*np.cos(x)
y5 = np.sin(x-0.5)

fig, axes = plt.subplots(nrows=1,ncols=2,figsize=(12,4),layout='constrained')
ax1 = axes[0]
ax2 = axes[1]

cs1 = ColorSet('vivid')
cs2 = ColorSet(shades_of_red)

cs1.apply_to_axes(ax1)
cs2.apply_to_axes(ax2)

for i in range(1,6):
    ax1.plot(x,i*y1)
    ax2.plot(x,i*y2)
```
